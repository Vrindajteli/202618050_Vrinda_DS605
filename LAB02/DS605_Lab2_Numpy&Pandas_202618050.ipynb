### Task 2

```python
# Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().
```

```python
matrix_A = np.array([
    [4, 7, 2],
    [3, 5, 1],
    [2, 1, 6]
])

matrix_B = np.array([
    [1, 0, 2],
    [5, 3, 1],
    [2, 4, 3]
])
```

```python
# Addition (Element-wise)
add_result = matrix_A + matrix_B
print("Matrix Addition (A + B):\n", add_result, "\n")
```

```python
# Element-wise Multiplication
elem_mult_result = matrix_A * matrix_B
print("Element-wise Multiplication (A * B):\n", elem_mult_result, "\n")
```

```python
# Matrix Multiplication (Dot Product) using @
matmul_result = matrix_A @ matrix_B
# Equivalent to: np.matmul(matrix_A, matrix_B)
print("Matrix Multiplication (A @ B):\n", matmul_result, "\n")
```

```python
# 2. Square Matrix Analysis (Matrix A)

# Transpose
A_transpose = matrix_A.T
print("Transpose of Matrix A (A.T):\n", A_transpose, "\n")

# Determinant
det_A = np.linalg.det(matrix_A)
print(f"Determinant of Matrix A: {det_A:.4f}\n")
```

```python
# 3. Invertibility Check & Verification
if not np.isclose(det_A, 0):
    # Compute inverse
    A_inv = np.linalg.inv(matrix_A)
    print("Inverse of Matrix A (A^-1):\n", A_inv, "\n")

    # Compute product A @ A^-1
    identity_check = matrix_A @ A_inv
    print("Product (A @ A^-1):\n", identity_check, "\n")

    # Verify equality using np.allclose()
    identity_true = np.eye(matrix_A.shape[0])
    is_identity = np.allclose(identity_check, identity_true)

    print(f"Verification (A @ A^-1 == I): {is_identity}")
else:
    print("Matrix A is singular (determinant is 0) and cannot be inverted.")
```

### Task 3

```python
# 1. Set seed and parameters
np.random.seed(42)
true_mean = 50.0
true_std = 10.0
sample_size = 10000
```

```python
# Generate normal distribution sample
data = np.random.normal(loc=true_mean, scale=true_std, size=sample_size)
```

```python
# 2. Compute sample mean and sample std
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)
```

```python
print(f"Chosen Mean: {true_mean}")
print(f"Sample Mean: {sample_mean:.4f}")
print(f"Difference (Mean): {abs(sample_mean - true_mean):.4f}\n\n")
print(f"Chosen Std: {true_std}")
print(f"Sample Std: {sample_std:.4f}")
print(f"Difference (Std): {abs(sample_std - true_std):.4f}")
```

```python
# 3. Plot histogram
plt.figure(figsize=(8, 5))
count, bins, ignored = plt.hist(data, bins=40, density=True, alpha=0.6, color='skyblue', edgecolor='black', label='Histogram (Sample Data)')

# Plot theoretical probability density function (PDF) curve
pdf = (1 / (true_std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((bins - true_mean) / true_std)**2)
plt.plot(bins, pdf, linewidth=2, color='darkred', label=f'Theoretical PDF (μ={true_mean}, σ={true_std})')

plt.axvline(sample_mean, color='blue', linestyle='dashed', linewidth=1.5, label=f'Sample Mean ({sample_mean:.2f})')

plt.title('Histogram of Normal Distribution Sample vs Theoretical PDF')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('normal_distribution_histogram.png', dpi=300)
plt.show()
```

### Task 5 - Filtering and Querying

```python
# 1. Male passengers older than 50
males_over_50 = len(df.query("Sex == 'male' and Age > 50"))
print(f"Male passengers older than 50: {males_over_50}")
```

```python
# 2. Female first-class passengers and survival percentage
female_p1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
female_p1_count = len(female_p1)
female_p1_surv_pct = (female_p1['Survived'].mean()) * 100
print(f"Female 1st-class passengers: {female_p1_count}")
print(f"Female 1st-class survival rate: {female_p1_surv_pct:.2f}%")
```

```python
# 3. Age 20-40, Fare above overall median, and survived
overall_median_fare = df['Fare'].median()
q3_count = len(df.query("20 <= Age <= 40 and Fare > @overall_median_fare and Survived == 1"))
print(f"Age 20-40 with Fare > overall median ({overall_median_fare:.2f}) who survived: {q3_count}")
```

```python
# 4. Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive
q4_count = len(df.query("SibSp == 0 and Parch == 0 and Age < 30 and Survived == 0"))
print(f"Alone, Age < 30, and did not survive: {q4_count}")
```

```python
# 5. Embarked='S', Pclass 2 or 3, and Fare above Southampton median
southampton_median_fare = df[df['Embarked'] == 'S']['Fare'].median()
q5_count = len(df.query("Embarked == 'S' and Pclass in [2, 3] and Fare > @southampton_median_fare"))
print(f"Embarked 'S', Pclass 2/3, Fare > Southampton median ({southampton_median_fare:.2f}): {q5_count}")
```

### Task 7 — Missing Values and Fare Outliers

```python
# 1. Missing value count and percentage
missing_count = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing_count, 'Percentage (%)': missing_pct})
print("Missing Values")
print(missing_df[missing_df['Missing Count'] > 0])
```

```python
# Plot missing counts as a bar chart
plt.figure(figsize=(7, 4))
missing_count[missing_count > 0].plot(kind='bar', color='coral', edgecolor='black')
plt.title('Missing Value Counts per Column')
plt.ylabel('Count')
plt.xlabel('Columns')
plt.tight_layout()
plt.show()
```

```python
# 2. Fill missing Age with mean Age (Show before/after)
print(f"Age missing count BEFORE fill: {df['Age'].isnull().sum()}")
df['Age_Mean_Filled'] = df['Age'].fillna(df['Age'].mean())
print(f"Age missing count AFTER mean fill: {df['Age_Mean_Filled'].isnull().sum()}")
```

```python
# 3. Four Imputation Strategies on Age (Mean, Median, Mode, Random Sampling)
df['Age_Impute_Mean'] = df['Age'].fillna(df['Age'].mean())
df['Age_Impute_Median'] = df['Age'].fillna(df['Age'].median())
df['Age_Impute_Mode'] = df['Age'].fillna(df['Age'].mode()[0])
```

```python
# Random Value Imputation (sampling from existing non-null age values)
random_ages = df['Age'].dropna().sample(df['Age'].isnull().sum(), random_state=42)
random_ages.index = df[df['Age'].isnull()].index
df['Age_Impute_Random'] = df['Age'].fillna(random_ages)
```

```python
# 4. Fare Outlier Detection via IQR
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]
```

```python
print("Fare Outlier Analysis")
print(f"Q1 (25th percentile): {Q1:.2f}")
print(f"Q3 (75th percentile): {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
print(f"Total Outliers Count: {len(outliers)}")
```

### Task 8 — Features and Pivot Table

```python
# 1. Feature Engineering
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
```

```python
# 2. Pivot Table: rows=Sex, columns=Pclass, values=mean Survived
pivot_tbl = df.pivot_table(index='Sex', columns='Pclass', values='Survived', aggfunc='mean')
```

```python
print("     Pivot Table: Survival Rate   ")
print(pivot_tbl)
```

```python
# Find highest and lowest survival groups
highest_group = pivot_tbl.unstack().idxmax()
highest_val = pivot_tbl.unstack().max()
lowest_group = pivot_tbl.unstack().idxmin()
lowest_val = pivot_tbl.unstack().min()

print(f"Highest Survival Group: Pclass {highest_group[0]}, Sex: {highest_group[1]} ({highest_val*100:.2f}%)")
print(f"Lowest Survival Group:  Pclass {lowest_group[0]}, Sex: {lowest_group[1]} ({lowest_val*100:.2f}%)")
```
